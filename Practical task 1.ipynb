```python
plt.hist(df["Purchase Amount (USD)"])
plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
df["Category"].value_counts().plot(kind="bar")
plt.title("Product Category Analysis")
plt.xlabel("Product Category")
plt.ylabel("Number of Sales")
plt.show()
```

```python
df["Location"].value_counts().plot(kind="bar")
plt.title("Store Location Analysis")
plt.xlabel("Store Location")
plt.ylabel("Number of Sales")
plt.show()
```

```python
df["Frequency of Purchases"].value_counts().plot(kind="bar")
plt.title("Frequency of Purchases")
plt.xlabel("Day")
plt.ylabel("Number of purchases")
plt.show()
```

```python
df["Discount Applied"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Discount Applied")
plt.ylabel("")
plt.show()
```

```python
plt.hist(df["Purchase Amount (USD)"], bins=10)
plt.title("Purchase Amount (USD) Distribution")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Subscription Status"], bins=10)
plt.title("Subscription Status")
plt.xlabel("Subscription Status")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["Review Rating"], df["Purchase Amount (USD)"])
plt.title("Review Rating vs Purchase Amount")
plt.xlabel("Review Rating")
plt.ylabel("Purchase Amount (USD)")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Oranges")
plt.title("Correlation Heatmap")
plt.show()
```
